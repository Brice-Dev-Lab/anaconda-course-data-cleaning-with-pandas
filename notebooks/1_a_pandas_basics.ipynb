There are two main types of data structures in Pandas: the `Series` and the `DataFrame`.  

## Series
A `Series` holds a single dimension of data.  Think of it as a single column, similar to a plain `list` in Python. But there are some key features that Pandas offers. 

```python
# declare a Series of temperatures
my_series = pd.Series([97.5, 98.2, 99.5, 98.1, 98.7], name='temperatures')

# print the series
my_series
```

Note how we called the `Series()` construct in Pandas, and passed a Python list of floating point values to it.  We gave it a name "temperatures."  When we print the series, we see not just the data but also the name provided and the inferred data type (`float54`, which is 64-bit floating point values).  This shows we can store useful metadata with our list of values.

But notice how there ar two columns, not just one, in our series.  The second column is the **index**

```python
# declare a Series of temperatures with a defined index

my_series = pd.Series(
    index = ['MON', 'TUES', 'WED', 'THURS', 'FRI'],
    data = [97.5, 98.2, 99.5, 98.1, 98.7],
    name = 'temperatures')

# pring the series
my_series
```

```python
# print the wed element

my_series['WED']
```

We aso call index the **axis**, and we call each element **axis labels** (e.g. consecutive integers or MON, TUE, WED...).  We call the actual data the **values** in the series.  The values do not have to be numeric.  They can be strings, dates, or any arbitrary Python type.  It is ideal a given series has the same datatype throughout its values for efficiency, and therefore can leverage fast vectorized operations behind the scenes.

## DataFrames
Whereas a series holds a single column of data, a **dataframe** in Pandas holds multiple columns of data.  It is also the data structure you will work with the most in Python.

```python
df = pd.DataFrame(data=[
    ('Otter Antiques', '5732 Serenity Ln', 'Addison', 'TX', 75031, 'COMMERCIAL', 55),
    ('North Exa Energy', '67 Hearst Dr', 'Plano', 'TX', 75093, 'INDUSTRIAL', 73),
    ('City of Plano', '239 Plano Dr', 'Plano', 'TX', 75093, 'GOVERNMENT', 193),
    ],
    columns=['COMPANY_NAME', 'ADDRESS', 'CITY', 'STATE', 'ZIP', 'CATEGORY', 'NUM_EMPLOYEES']
)

# print df
df
```

We can initialize thes three recods by assigning a collection containing collections to the `data` paramet, where each sub-collection is a row of data.  In this case I create a list of tuples.  The names of each column are provided as a list of strings to the `columns` parament.  Notice there are seven colunns and three rows.  We can obtain thes dimensions using the dataframe's `shape` property.

```python
df.shape
```

```python
df['ADDRESS']
```

```python
# NOTE:  for multiple columns, a double bracket must be used
df[['ADDRESS', 'COMPANY_NAME']]
```

```python
df.columns
```

```python
df.index
```

Note the index implementation returned is a `RangeIndex` which is a simple integer index defined by a range and step (typically `1` for a simple consecutive integer).  To retrieve an index by row, use the `loc[]` getter with the row number.

```python
df.loc[2]
```

You can manipulate and change the row index, most commonly by using the `set_index()` function and specifying which column you want as the index.  Below we set the 'CATEGORY' column to become the index.  Note we set `inplace` to be true so it edits the existing datagrame rather than create a new one.

```python
df.set_index('CATEGORY', inplace=True)
```

You can use the `loc[]` getter to look up a record by the company name

```python
df.loc['COMMERCIAL']
```

If you study the `index` property now, notice it is no longer a `RangeIndex` but rather a plain `Index` organized by string objects.

```python
df.index
```

To reset the index back to the default, call `reset_index()`. Use the `inplace`modifier so it edits the existing dataframe rather than create a new one.

```python
df.reset_index(inplace=True)
```
